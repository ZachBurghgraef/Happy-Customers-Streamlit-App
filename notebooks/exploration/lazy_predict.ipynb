# Using lazypredict to scan many models before digging in and exploring the best models

___________________________

```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np
import sklearn
import lazypredict
```

```python
# importing the data into data frame
filename = r"../../data/raw/ACME-HappinessSurvey2020.csv"
df = pd.read_csv(filename)
df.describe()
```

```python
from lazypredict.Supervised import LazyClassifier
from sklearn.model_selection import train_test_split

# format data
X = df.copy()
Y = X.Y
X = X.drop(["Y"], axis=1)

# Split into train and hold-out test
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
clf = LazyClassifier(verbose=0, ignore_warnings=True, custom_metric=None)
models, predictions = clf.fit(X_train, X_test, y_train, y_test)
```
